## Loading and Processing the Dataset

```python
!kaggle datasets download -d orvile/brain-cancer-mri-dataset --unzip
```

```python
from datasets import load_dataset

data_dir = "../data\Brain_Cancer"

# Define proportions for train and validation splits
train_size = 0.8  
validation_size = 0.2  

#Load the dataset
data = load_dataset("imagefolder", data_dir=data_dir, split="train")#divide en entrenamiento y test

# Split the dataset into train and validation sets
data = data.train_test_split(
    train_size=train_size,
    test_size=validation_size,
    shuffle=True,
    seed=42,
)

# Rename the 'test' split to 'validation'
data["validation"] = data.pop("test")

# Display dataset details
print(data)
```
